```python
import pandas as pd
```

```python
## Carregando a base
df_alunos = pd.read_csv("PDA_2022-2024_1.2_Alunos_Anonimo.csv",
                        sep=';',encoding='cp1252')
df_alunos.info()
```

```python
## Renomeando colunas
df_alunos = df_alunos.rename(columns=df_alunos.iloc[1])
df_alunos = df_alunos.drop([0,1]).reset_index(drop=True)
df_alunos.head()
```

```python
## Verificando valores das colunas
df_alunos.describe()
```

```python
### Criando um Df para cada coluna, a fim de analisar os valores únicos de cada coluna

nome_colunas = df_alunos.columns.to_list()

nome_df = nome_colunas.copy()
for nome in nome_df:
    if r'/' in nome:
        i = nome_df.index(nome)
        nome_df[i] = nome_df[i].replace(r"/", "_")
```

```python
dfs = {}

for coluna in df_alunos.columns:
    dfs[coluna] = (
        df_alunos[coluna]
        .value_counts(dropna=False)
        .reset_index()
    )
    dfs[coluna].columns = [coluna, 'Contagem']
```

```python
## Contagem valores com '/'

contagem_dados_inconsistentes = 0
contagem_dados = df_alunos.shape[0]

for i in df_alunos.index:
    valor = df_alunos['Período de Ingresso'][i]

    if r'/' in valor:
        contagem_dados_inconsistentes+=1

percentual_inconsistencia = (contagem_dados_inconsistentes/contagem_dados) * 100

print(f'Contagem de dados inconsistentes: {contagem_dados_inconsistentes}')
print(f'Percentual de dados inconsistentes: {percentual_inconsistencia:.2f}%')
```

É possível excluir os registros que possuem valores *inconsistentes*, pois equivale a somente 7.50% dos dados totais.

```python
dfs['Situação de Matrícula']
```

```python
df = df_alunos['Situação de Matrícula'].unique().tolist()
filtro = [nome for nome in df if '-' not in nome and r'/' not in nome]
filtro
```

```python
contagem_dados_inconsistentes = 0
contagem_dados = df_alunos.shape[0]

for i in df_alunos.index:
    valor  = df_alunos['Situação de Matrícula'][i]
    if valor not in filtro:
        contagem_dados_inconsistentes += 1

percentual_inconsistencia = (contagem_dados_inconsistentes/contagem_dados) * 100

print(f'Total de dados da amostra: {contagem_dados}')
print(f'Percentual de dados incosistentes {percentual_inconsistencia:.2f}%')
```

```python
df_dados_inconsistentes = df_alunos[~df_alunos['Situação de Matrícula'].isin(filtro)]
df_dados_inconsistentes.describe()
```

Alterar valores da coluna *Situação Matrícula*, dos alunos *Concluinte* para *Formando*.

Aplicando edições no df original.

```python
## Selecionando as colunas pertinentes
df_dados_editados = df_alunos.iloc[:,[0,2,3,4,5,6,7]]

## Alterando o valor das linhas com valores inconsistentes na coluna Situação de Matrícula
df_dados_editados.loc[
    ~df_dados_editados['Situação de Matrícula'].isin(filtro),
    'Situação de Matrícula'   
] = "FORMANDO"


## Apagar linhas com valores inconsistentes na coluna Período de Ingresso
df_dados_editados = df_dados_editados.iloc[~df_dados_editados['Período de Ingresso'].str.contains(r'/',regex=False)].reset_index(drop=True)

## Imprimindo DF
df_dados_editados
```

```python
## Exportando Dados Editados CSV
df_dados_editados.to_csv('dados_editados.csv', index=False)
```
